# 📏 C. Confidence Interval

**Tujuan Pembelajaran 1** — Mengestimasi rata-rata tip populasi.

$$CI = \bar{x} \pm t_{\alpha/2,\, n-1} \cdot \frac{s}{\sqrt{n}}$$

## C.1 CI 95% Manual vs scipy


```python
x = df['tip']
n, xbar, s = len(x), x.mean(), x.std(ddof=1)
se = s / np.sqrt(n)
t_crit = stats.t.ppf(1 - ALPHA/2, df=n-1)
margin = t_crit * se

print("=== CI 95% RATA-RATA TIP (MANUAL) ===")
print(f"  n          = {n}")
print(f"  x̄          = ${xbar:.3f}")
print(f"  s          = ${s:.3f}")
print(f"  SE = s/√n  = {se:.4f}")
print(f"  t kritis   = {t_crit:.4f}  (df = {n-1})")
print(f"  margin     = {margin:.4f}")
print(f"  CI 95%     = [${xbar-margin:.3f}, ${xbar+margin:.3f}]")

ci_scipy = stats.t.interval(0.95, df=n-1, loc=xbar, scale=se)
print(f"\nVerifikasi scipy: [${ci_scipy[0]:.3f}, ${ci_scipy[1]:.3f}] ✅")
```

```python
# Pengaruh tingkat kepercayaan terhadap lebar CI
print(f"{'Level':>7} | {'t kritis':>8} | {'CI':>22} | {'Lebar':>6}")
print("-" * 54)
for conf in [0.80, 0.90, 0.95, 0.99]:
    lo, hi = stats.t.interval(conf, df=n-1, loc=xbar, scale=se)
    print(f"{conf*100:>6.0f}% | {stats.t.ppf((1+conf)/2, n-1):>8.3f} | [${lo:.3f}, ${hi:.3f}]   | {hi-lo:.3f}")
print("\n💡 Makin tinggi tingkat kepercayaan → makin LEBAR interval (trade-off presisi vs keyakinan).")
```

## C.2 Simulasi: Apa Arti "95%" Sebenarnya?

Kita anggap 244 transaksi sebagai **populasi** (μ diketahui). Ambil 100 sampel acak berukuran 30, hitung CI 95% untuk masing-masing, lalu hitung berapa interval yang mengandung μ.


```python
mu_pop = df['tip'].mean()
n_sim, n_sample = 100, 30
cis, hits = [], 0
for _ in range(n_sim):
    smp = df['tip'].sample(n_sample, replace=False)
    lo, hi = stats.t.interval(0.95, df=n_sample-1, loc=smp.mean(), scale=smp.std(ddof=1)/np.sqrt(n_sample))
    covered = lo <= mu_pop <= hi
    hits += covered
    cis.append((lo, hi, covered))

fig, ax = plt.subplots(figsize=(12, 6))
for i, (lo, hi, ok) in enumerate(cis):
    ax.plot([lo, hi], [i, i], color='#059669' if ok else '#F43F5E', linewidth=2 if ok else 3)
ax.axvline(mu_pop, color='black', linestyle='--', linewidth=2, label=f'μ populasi = ${mu_pop:.2f}')
ax.set_xlabel('Tip ($)'); ax.set_ylabel('Sampel ke-')
ax.set_title(f'100 CI 95% dari sampel n=30 → {hits} mengandung μ (merah = meleset)', fontweight='bold')
ax.legend(); plt.tight_layout(); plt.show()

print(f"Coverage empiris = {hits}/{n_sim} = {hits}%  (teori ≈ 95%)")
print("💡 Inilah makna 95%: sifat METODE, bukan peluang satu interval tertentu.")
```

# 🧮 D. t-test

**Tujuan Pembelajaran 2 & 3** — Kerangka 5 langkah.

## D.1 One-sample t-test: Apakah rata-rata tip = standar industri 15%?

- **H₀**: μ_tip_pct = 15%
- **H₁**: μ_tip_pct ≠ 15%
- **α** = 0.05


```python
t_stat, p_val = stats.ttest_1samp(df['tip_pct'], popmean=15)
print("=== ONE-SAMPLE T-TEST: tip_pct vs 15% ===")
print(f"  Mean tip_pct = {df['tip_pct'].mean():.2f}%")
print(f"  t = {t_stat:.3f},  p = {p_val:.4f}")
print(f"  Keputusan: {'TOLAK H₀' if p_val < ALPHA else 'GAGAL TOLAK H₀'}")
if p_val < ALPHA:
    print("  → Rata-rata tip% berbeda signifikan dari 15%.")
else:
    print("  → Tidak cukup bukti bahwa rata-rata tip% berbeda dari 15%.")
```

## D.2 Two-sample t-test: Apakah tip pria ≠ tip wanita?

- **H₀**: μ_pria = μ_wanita
- **H₁**: μ_pria ≠ μ_wanita

**Langkah 3 — cek asumsi**: normalitas (Shapiro) dan kesamaan varians (Levene). Jika varians berbeda, pakai **Welch's t-test** (`equal_var=False`).


```python
male, female = df.loc[df.sex == 'Male', 'tip'], df.loc[df.sex == 'Female', 'tip']

print("=== CEK ASUMSI ===")
for name, g in [('Male', male), ('Female', female)]:
    w, p = stats.shapiro(g)
    print(f"  Shapiro {name:<6}: W={w:.3f}, p={p:.4f} → {'normal' if p > ALPHA else 'TIDAK normal'}")
lev_stat, lev_p = stats.levene(male, female)
print(f"  Levene         : p={lev_p:.4f} → varians {'sama' if lev_p > ALPHA else 'BERBEDA'}")
print("  💡 n > 30 per grup → CLT membuat t-test tetap robust meski data tidak normal.")

t_stat, p_val = stats.ttest_ind(male, female, equal_var=False)
pooled_sd = np.sqrt(((len(male)-1)*male.var() + (len(female)-1)*female.var()) / (len(male)+len(female)-2))
cohens_d = (male.mean() - female.mean()) / pooled_sd

print("\n=== WELCH T-TEST: tip Male vs Female ===")
print(f"  Mean Male   = ${male.mean():.3f} (n={len(male)})")
print(f"  Mean Female = ${female.mean():.3f} (n={len(female)})")
print(f"  Selisih     = ${male.mean()-female.mean():.3f}")
print(f"  t = {t_stat:.3f},  p = {p_val:.4f}")
print(f"  Cohen's d   = {cohens_d:.3f}  (|d|<0.2 sangat kecil, 0.5 sedang, 0.8 besar)")
print(f"  Keputusan   : {'TOLAK H₀' if p_val < ALPHA else 'GAGAL TOLAK H₀'}")
print("\n💡 Grafik tampak beda, tapi bukti statistik belum cukup. Efeknya pun kecil.")
```

# 📊 E. Chi-Square Test of Independence

Untuk dua variabel **kategorik**. Syarat: frekuensi harapan (expected) tiap sel ≥ 5.

## E.1 Smoker × Time
- **H₀**: status merokok independen dari waktu makan
- **H₁**: ada hubungan


```python
def chi_square_report(var1, var2):
    ct = pd.crosstab(df[var1], df[var2])
    chi2, p, dof, expected = stats.chi2_contingency(ct)
    n = ct.values.sum()
    cramers_v = np.sqrt(chi2 / (n * (min(ct.shape) - 1)))
    print(f"=== CHI-SQUARE: {var1} × {var2} ===")
    print("Observed:"); print(ct)
    print(f"\nExpected min = {expected.min():.1f}  ({'OK ≥5' if expected.min() >= 5 else 'PERINGATAN <5'})")
    print(f"χ² = {chi2:.3f}, dof = {dof}, p = {p:.4g}")
    print(f"Cramér's V = {cramers_v:.3f}  (0.1 lemah, 0.3 sedang, 0.5 kuat)")
    print(f"Keputusan: {'TOLAK H₀ — ada hubungan' if p < ALPHA else 'GAGAL TOLAK H₀ — tidak ada bukti hubungan'}\n")
    return p

p_smoker_time = chi_square_report('smoker', 'time')
```

## E.2 Day × Time
- **H₀**: hari independen dari waktu makan


```python
p_day_time = chi_square_report('day', 'time')
print("💡 Hampir semua transaksi Thur = Lunch, Sat/Sun = Dinner → hubungan SANGAT kuat.")
print("   Catatan: ada sel expected < 5 — pada laporan nyata, gabungkan kategori atau pakai Fisher's exact.")
```

# 📈 F. One-way ANOVA + Post-hoc Tukey

Membandingkan mean **≥3 grup** sekaligus. Mengapa tidak t-test berulang? 6 pasang t-test → risiko Error Tipe I membengkak (ingat Bonferroni di M5).

- **H₀**: μ_Thur = μ_Fri = μ_Sat = μ_Sun
- **H₁**: minimal satu hari berbeda


```python
order = ['Thur', 'Fri', 'Sat', 'Sun']
groups = [df.loc[df.day == d, 'total_bill'] for d in order]

print("Mean total_bill per hari:")
print(df.groupby('day')['total_bill'].agg(['mean', 'std', 'count']).loc[order].round(2))

lev_p = stats.levene(*groups).pvalue
f_stat, p_val = stats.f_oneway(*groups)
grand = df['total_bill'].mean()
ss_between = sum(len(g) * (g.mean() - grand)**2 for g in groups)
ss_total = ((df['total_bill'] - grand)**2).sum()
eta_sq = ss_between / ss_total

print(f"\nLevene p = {lev_p:.4f}")
print(f"=== ONE-WAY ANOVA: total_bill ~ day ===")
print(f"  F = {f_stat:.3f},  p = {p_val:.4f}")
print(f"  η² (eta squared) = {eta_sq:.3f}  → {eta_sq*100:.1f}% variasi bill dijelaskan oleh hari")
print(f"  Keputusan: {'TOLAK H₀ — minimal satu hari berbeda' if p_val < ALPHA else 'GAGAL TOLAK H₀'}")
```

```python
# ANOVA hanya bilang "ada yang beda" — Tukey HSD memberi tahu PASANGAN mana
tukey = pairwise_tukeyhsd(endog=df['total_bill'], groups=df['day'], alpha=ALPHA)
print(tukey.summary())
print("\n💡 Kolom 'reject' = True menandai pasangan yang berbeda signifikan.")
print("   Perhatikan: ANOVA signifikan, tapi bisa saja tidak ada pasangan yang lolos Tukey —")
print("   ini terjadi saat efek tersebar tipis di beberapa pasangan (p ANOVA dekat 0.05).")
```

# 📝 H. Template Milestone 2 + Refleksi

## H.1 Template Uji Hipotesis untuk Dataset Anda

Copy cell di bawah ke **notebook Milestone 1 Anda** (dataset Kaggle yang sama), lalu isi 2-3 hipotesis yang berasal dari insight EDA.


```python
# ==========================================
# MILESTONE 2 - HYPOTHESIS TESTING
# Dataset: Google Play Store Apps
# ==========================================

import pandas as pd
import numpy as np
from scipy import stats
from IPython.display import display

# Membaca dataset
df_my = pd.read_csv("googleplaystore.csv")

# Membersihkan kolom numerik
df_my["Rating"] = pd.to_numeric(df_my["Rating"], errors="coerce")

df_my["Reviews"] = pd.to_numeric(
    df_my["Reviews"], errors="coerce"
)

df_my["Installs"] = pd.to_numeric(
    df_my["Installs"].astype(str)
    .str.replace(",", "", regex=False)
    .str.replace("+", "", regex=False),
    errors="coerce"
)

df_my["Price"] = pd.to_numeric(
    df_my["Price"].astype(str)
    .str.replace("$", "", regex=False),
    errors="coerce"
)

# Membuat data rating valid (skala 1-5)
df_rating = df_my[
    df_my["Rating"].between(1, 5)
].copy()

# Membuat kelompok aplikasi gratis dan berbayar
df_my["Kelompok_Harga"] = pd.Series(
    np.select(
        [
            df_my["Price"].eq(0),
            df_my["Price"].gt(0)
        ],
        [
            "Gratis",
            "Berbayar"
        ],
        default=None
    ),
    index=df_my.index,
    dtype="object"
)

print("Jumlah data awal:", len(df_my))
print("Jumlah data dengan rating valid:", len(df_rating))
print("Jumlah aplikasi gratis:",
      (df_my["Kelompok_Harga"] == "Gratis").sum())
print("Jumlah aplikasi berbayar:",
      (df_my["Kelompok_Harga"] == "Berbayar").sum())
```

```python
# ==========================================
# CONFIDENCE INTERVAL 95%
# ==========================================

alpha = 0.05

x = df_rating["Rating"].dropna()

n = len(x)
mean_rating = x.mean()
std_rating = x.std(ddof=1)
standard_error = stats.sem(x)

ci_lower, ci_upper = stats.t.interval(
    confidence=0.95,
    df=n-1,
    loc=mean_rating,
    scale=standard_error
)

print("CONFIDENCE INTERVAL 95%")
print("Jumlah sampel:", n)
print("Rata-rata Rating:", round(mean_rating, 4))
print("Standar Deviasi:", round(std_rating, 4))
print("Standard Error:", round(standard_error, 4))
print("Batas bawah CI:", round(ci_lower, 4))
print("Batas atas CI:", round(ci_upper, 4))

print(
    f"\nDengan tingkat kepercayaan 95%, "
    f"rata-rata rating populasi diperkirakan berada "
    f"antara {ci_lower:.4f} sampai {ci_upper:.4f}."
)
```

```python
# ==========================================
# HIPOTESIS 1
# REVIEWS VS INSTALLS
# ==========================================

alpha = 0.05

data_h1 = df_my[
    (df_my["Reviews"] > 0) &
    (df_my["Installs"] > 0)
][["Reviews", "Installs"]].dropna().copy()

# Transformasi logaritmik
data_h1["Log_Reviews"] = np.log10(data_h1["Reviews"])
data_h1["Log_Installs"] = np.log10(data_h1["Installs"])

x1 = data_h1["Log_Reviews"]
y1 = data_h1["Log_Installs"]

# Cek asumsi dan pola hubungan
correlation, p_value = stats.pearsonr(x1, y1)
hasil_h1 = {
    "Statistik Uji": correlation,
    "p-value": p_value,
    "Effect Size": abs(correlation)
}

print("HIPOTESIS 1: REVIEWS DAN INSTALLS")
print("H0: Tidak terdapat hubungan signifikan.")
print("H1: Terdapat hubungan signifikan.")
print("Alpha:", alpha)

print("\nCEK ASUMSI")
print("Jumlah sampel:", len(data_h1))
print("Korelasi Pearson digunakan untuk menguji hubungan linear.")
print("Periksa scatter plot untuk melihat linearitas dan outlier.")

# Scatter plot
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))
plt.scatter(x1, y1, alpha=0.3)
plt.xlabel("Log10 Reviews")
plt.ylabel("Log10 Installs")
plt.title("Hubungan Reviews dan Installs")
plt.show()

# Hasil uji
print("\nHASIL PENGUJIAN")
print("Koefisien Pearson (r):", round(correlation, 4))
print("p-value:", p_value)

# Effect size
effect_size = abs(correlation)

if effect_size < 0.1:
    kategori = "Sangat kecil"
elif effect_size < 0.3:
    kategori = "Kecil"
elif effect_size < 0.5:
    kategori = "Sedang"
else:
    kategori = "Besar"

print("Effect Size:", round(effect_size, 4), "-", kategori)

# Kesimpulan statistik dan bisnis
if p_value < alpha:
    print("\nKESIMPULAN STATISTIK: H0 ditolak.")
    print("Terdapat hubungan yang signifikan antara Reviews dan Installs.")
    print("KESIMPULAN BISNIS: Jumlah ulasan aplikasi memiliki hubungan")
    print("dengan jumlah instalasi. Namun, korelasi tidak membuktikan sebab-akibat.")
else:
    print("\nKESIMPULAN STATISTIK: H0 tidak dapat ditolak.")
    print("Tidak ditemukan hubungan yang signifikan antara Reviews dan Installs.")
    print("KESIMPULAN BISNIS: Jumlah ulasan belum terbukti memiliki hubungan")
    print("yang signifikan dengan jumlah instalasi pada data ini.")
```

```python
# ==========================================
# HIPOTESIS 2
# RATING APLIKASI GRATIS VS BERBAYAR
# ==========================================

alpha = 0.05

gratis = df_my[
    (df_my["Kelompok_Harga"] == "Gratis") &
    (df_my["Rating"].between(1, 5))
]["Rating"].dropna()

berbayar = df_my[
    (df_my["Kelompok_Harga"] == "Berbayar") &
    (df_my["Rating"].between(1, 5))
]["Rating"].dropna()

print("HIPOTESIS 2: RATING GRATIS VS BERBAYAR")
print("H0: Tidak terdapat perbedaan rata-rata rating.")
print("H1: Terdapat perbedaan rata-rata rating.")
print("Alpha:", alpha)

print("\nCEK ASUMSI")

# Uji normalitas Shapiro menggunakan sampel maksimal 5000
sampel_gratis = gratis.sample(
    n=min(5000, len(gratis)),
    random_state=42
)

sampel_berbayar = berbayar.sample(
    n=min(5000, len(berbayar)),
    random_state=42
)

shapiro_gratis = stats.shapiro(sampel_gratis)
shapiro_berbayar = stats.shapiro(sampel_berbayar)

print("Shapiro Gratis p-value:",
      round(shapiro_gratis.pvalue, 4))

print("Shapiro Berbayar p-value:",
      round(shapiro_berbayar.pvalue, 4))

# Uji kesamaan varians
levene_result = stats.levene(gratis, berbayar)

print("Levene p-value:",
      round(levene_result.pvalue, 4))

print("\nRata-rata Gratis:", round(gratis.mean(), 4))
print("Rata-rata Berbayar:", round(berbayar.mean(), 4))

# Welch's t-test
t_stat, p_value = stats.ttest_ind(
    gratis,
    berbayar,
    equal_var=False
)

print("\nHASIL PENGUJIAN")
print("Statistik t:", round(t_stat, 4))
print("p-value:", p_value)

# Effect size Cohen's d
n1 = len(gratis)
n2 = len(berbayar)

pooled_std = np.sqrt(
    ((n1 - 1) * gratis.var(ddof=1) +
     (n2 - 1) * berbayar.var(ddof=1)) /
    (n1 + n2 - 2)
)

cohens_d = (gratis.mean() - berbayar.mean()) / pooled_std
hasil_h2 = {
    "Statistik Uji": t_stat,
    "p-value": p_value,
    "Effect Size": abs(cohens_d)
}

print("Cohen's d:", round(cohens_d, 4))

if abs(cohens_d) < 0.2:
    kategori = "Sangat kecil"
elif abs(cohens_d) < 0.5:
    kategori = "Kecil"
elif abs(cohens_d) < 0.8:
    kategori = "Sedang"
else:
    kategori = "Besar"

print("Kategori Effect Size:", kategori)

# Kesimpulan
if p_value < alpha:
    print("\nKESIMPULAN STATISTIK: H0 ditolak.")
    print("Terdapat perbedaan rata-rata rating yang signifikan.")
    print("KESIMPULAN BISNIS: Status gratis atau berbayar berkaitan")
    print("dengan perbedaan rating aplikasi pada dataset ini.")
else:
    print("\nKESIMPULAN STATISTIK: H0 tidak dapat ditolak.")
    print("Tidak ditemukan perbedaan rata-rata rating yang signifikan.")
    print("KESIMPULAN BISNIS: Belum ada bukti bahwa aplikasi gratis")
    print("dan berbayar memiliki rata-rata rating yang berbeda.")
```

```python
# ==========================================
# HIPOTESIS 3
# RATING VS INSTALLS
# ==========================================

alpha = 0.05

data_h3 = df_my[
    df_my["Rating"].between(1, 5) &
    (df_my["Installs"] > 0)
][["Rating", "Installs"]].dropna().copy()

data_h3["Log_Installs"] = np.log10(
    data_h3["Installs"]
)

x3 = data_h3["Rating"]
y3 = data_h3["Log_Installs"]

print("HIPOTESIS 3: RATING DAN INSTALLS")
print("H0: Tidak terdapat hubungan signifikan.")
print("H1: Terdapat hubungan signifikan.")
print("Alpha:", alpha)

# Cek pola hubungan
correlation, p_value = stats.pearsonr(x3, y3)
hasil_h3 = {
    "Statistik Uji": correlation,
    "p-value": p_value,
    "Effect Size": abs(correlation)
}

print("\nCEK ASUMSI")
print("Jumlah sampel:", len(data_h3))
print("Periksa scatter plot untuk melihat linearitas dan outlier.")

plt.figure(figsize=(8, 5))
plt.scatter(x3, y3, alpha=0.3)
plt.xlabel("Rating")
plt.ylabel("Log10 Installs")
plt.title("Hubungan Rating dan Installs")
plt.show()

# Hasil pengujian
print("\nHASIL PENGUJIAN")
print("Koefisien Pearson (r):", round(correlation, 4))
print("p-value:", p_value)

# Effect size
effect_size = abs(correlation)

if effect_size < 0.1:
    kategori = "Sangat kecil"
elif effect_size < 0.3:
    kategori = "Kecil"
elif effect_size < 0.5:
    kategori = "Sedang"
else:
    kategori = "Besar"

print("Effect Size:", round(effect_size, 4), "-", kategori)

# Kesimpulan
if p_value < alpha:
    print("\nKESIMPULAN STATISTIK: H0 ditolak.")
    print("Terdapat hubungan yang signifikan antara Rating dan Installs.")
    print("KESIMPULAN BISNIS: Rating aplikasi memiliki hubungan")
    print("dengan jumlah instalasi, tetapi tidak membuktikan sebab-akibat.")
else:
    print("\nKESIMPULAN STATISTIK: H0 tidak dapat ditolak.")
    print("Tidak ditemukan hubungan yang signifikan antara Rating dan Installs.")
    print("KESIMPULAN BISNIS: Rating belum terbukti memiliki hubungan")
    print("yang signifikan dengan jumlah instalasi pada data ini.")
```

### ✏️ Jawaban Anda:

**1.** Pada simulasi 100 Confidence Interval (CI), apabila terdapat beberapa interval yang tidak mengandung nilai rata-rata populasi (\(\mu\)), hal tersebut tidak berarti perhitungan yang dilakukan salah. Confidence Level 95% berarti bahwa jika pengambilan sampel dan pembuatan interval dilakukan berulang kali, sekitar 95% dari interval yang terbentuk akan mencakup nilai rata-rata populasi sebenarnya, sedangkan sekitar 5% interval lainnya mungkin tidak mencakup nilai tersebut. Jadi, interval yang tidak mencakup \(\mu\) merupakan hal yang wajar dan sesuai dengan konsep Confidence Interval 95%.

**2.** Jika dataset diperbesar menjadi 100.000 transaksi dengan selisih rata-rata tip yang tetap sama, kemungkinan p-value akan menjadi lebih kecil dan bisa saja menjadi signifikan secara statistik (p-value kurang dari 0,05). Hal ini karena semakin besar ukuran sampel, semakin tinggi kemampuan uji statistik dalam mendeteksi perbedaan yang kecil.

Namun, perbedaan tersebut belum tentu menjadi penting secara bisnis. Jika Cohen's d tetap kecil, maka perbedaan tip antara pria dan wanita masih tergolong kecil secara praktis. Artinya, meskipun hasilnya signifikan secara statistik, perbedaan tersebut belum tentu memberikan dampak yang cukup besar untuk dijadikan dasar keputusan bisnis.

**3.** Dalam deteksi transaksi fraud e-wallet dengan H₀ bahwa transaksi merupakan transaksi normal, Error Tipe I adalah ketika transaksi normal dianggap sebagai transaksi fraud, sedangkan Error Tipe II adalah ketika transaksi fraud dianggap sebagai transaksi normal.

Menurut saya, Error Tipe II dapat menjadi lebih mahal karena transaksi fraud yang tidak terdeteksi dapat menyebabkan kerugian finansial bagi perusahaan maupun pengguna. Namun, Error Tipe I juga memiliki biaya karena transaksi normal yang salah ditandai sebagai fraud dapat menghambat pengguna dalam melakukan transaksi.

Jika perusahaan lebih memprioritaskan pencegahan kerugian akibat fraud, maka nilai α dapat dibuat lebih besar agar sistem lebih sensitif dalam mendeteksi transaksi mencurigakan, sehingga risiko fraud yang tidak terdeteksi dapat dikurangi. Akan tetapi, peningkatan α juga dapat meningkatkan kemungkinan transaksi normal salah terdeteksi. Oleh karena itu, pemilihan α perlu mempertimbangkan keseimbangan antara kerugian fraud dan gangguan terhadap transaksi normal.

**4.** Membandingkan empat hari menggunakan enam kali t-test berpasangan berbahaya karena semakin banyak pengujian statistik yang dilakukan, semakin besar kemungkinan terjadinya Error Tipe I, yaitu menolak H₀ padahal H₀ sebenarnya benar. Jika setiap pengujian menggunakan α = 0,05, maka risiko kesalahan secara keseluruhan akan meningkat.

ANOVA lebih tepat digunakan untuk membandingkan rata-rata dari lebih dari dua kelompok secara bersamaan dalam satu pengujian. Dengan ANOVA, kita dapat mengetahui apakah terdapat perbedaan rata-rata yang signifikan di antara empat hari tanpa meningkatkan risiko Error Tipe I sebanyak pengujian t-test berulang.

Hal ini berkaitan dengan konsep Bonferroni di Modul 5, yaitu metode untuk mengendalikan risiko Error Tipe I pada pengujian berganda dengan membagi nilai α berdasarkan jumlah pengujian. Misalnya, jika terdapat enam pengujian dan α = 0,05, maka nilai α yang digunakan untuk setiap pengujian dapat menjadi \(0,05/6 = 0,0083\). Dengan demikian, risiko kesalahan secara keseluruhan dapat lebih dikendalikan.
